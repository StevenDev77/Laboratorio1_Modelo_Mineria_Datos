# FASE 2 SEMMA: Análisis Exploratorio de Datos (Explore)
Este notebook documenta la exploración estadística y visualización del dataset de vulnerabilidades en microservicios.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../fase_1_sample/dataset_vulnerabilidades.csv')
print(f'Total registros: {len(df)}')
df.head()
```

```
print('=== DISTRIBUCIÓN BINARIA (SEGURO VS VULNERABLE) ===')
print(df['is_vulnerable'].value_counts(normalize=True))

print('\n=== DISTRIBUCIÓN POR LENGUAJE ===')
print(df['language'].value_counts())

print('\n=== DISTRIBUCIÓN POR CATEGORÍA TÉCNICA ===')
print(df['category'].value_counts())
```
